# Oil Price Dataset Analysis

*Plot*

```python
#Oil Price Over Time (2007-2024)
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('combined_price_dataset.csv')

df['Date'] = pd.to_datetime(df['Date'])

plt.figure(figsize=(12,8))

plt.plot(df['Date'], df['Average Price'], label = 'Average Price', color='blue')
plt.plot(df['Date'], df['Brent Price'], label = 'Brent Price', color='red')
plt.plot(df['Date'], df['WTI Price'], label = 'WTI Price', color='green')

plt.title('Oil Price Over Time (2007-2024)', fontsize=16)
plt.xlabel('Date')
plt.ylabel('Oil Price')

plt.grid(True)
plt.legend()
plt.show()
```

*Normalization*

```python
#Oil Price Over Time (2007-2024)
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('combined_price_dataset_normalized.csv')

df['Date'] = pd.to_datetime(df['Date'], format='%Y-%m-%d')

plt.figure(figsize=(12,8))

plt.plot(df['Date'], df['Average Price'], label = 'Average Price', color='blue')
plt.plot(df['Date'], df['Brent Price'], label = 'Brent Price', color='red')
plt.plot(df['Date'], df['WTI Price'], label = 'WTI Price', color='green')

plt.title('Oil Price Over Time (2007-2024)', fontsize=16)
plt.xlabel('Date')
plt.ylabel('Oil Price')
plt.grid(True)
plt.legend()
plt.show()
```

*Corelation Analysis*

```python
#Corealtion Analysis of Prices
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('combined_price_dataset.csv')
columns = ['Average Price', 'Brent Price', 'WTI Price']

matrix = df[columns].corr()

sns.heatmap(matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title('Correlation Matrix Of Oil Prices')
plt.figure(figsize=(12,8))

plt.show()
```
