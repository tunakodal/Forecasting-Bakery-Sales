#Weather Dataset Analysis

*Plot*

```python
#Weather Data Over Time (2015-2024)
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('edinburgh_weather_data.csv')

df['time'] = pd.to_datetime(df['time'])
df['pres'] = df['pres'] / 1000 #mbar to bar

plt.figure(figsize=(12,8))
plt.title('Weather Data Over Time (2015-2024)', fontsize=16)
plt.xlabel('Date')
plt.ylabel('Weather Data')

plt.plot(df['time'], df['tmax'], label='Maximum Temprature', color='blue')
plt.plot(df['time'], df['tmin'], label='Minimum Temprature', color='yellow')
plt.plot(df['time'], df['tavg'], label='Average Temprature', color='green')
plt.plot(df['time'], df['wspd'], label='Wind Speed', color='red')
plt.plot(df['time'], df['pres'], label='Pressure (bar)', color='orange')
plt.legend()
plt.show()
```

*Corelation Anlaysis*

```python
#Corealtion Analysis of Weather Values
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('edinburgh_weather_data.csv')
columns = ['tmax','tmin','tavg','wspd','pres']

matrix = df[columns].corr()

sns.heatmap(matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title('Correlation Matrix Of Weather Values')
plt.figure(figsize=(12,8))

plt.show()
```
